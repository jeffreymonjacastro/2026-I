# Open midi file


```python
import mido
import numpy as np
import pygame
import time
```

```python
def open_midi(file_path):
    mid = mido.MidiFile(file_path)

    notes = []
    for track in mid.tracks:
        absolute_time = 0
        for msg in track:
            absolute_time += msg.time
            if msg.type == "note_on" and msg.velocity > 0:
                notes.append([absolute_time, msg.note, msg.velocity, msg.channel])

    notes_array = np.array(notes, dtype=np.int32)
    return notes_array
```

```python
def play_midi(file_path):
    pygame.mixer.init()
    pygame.mixer.music.load(file_path)
    pygame.mixer.music.play()

    try:
        while pygame.mixer.music.get_busy():
            time.sleep(1)
    except KeyboardInterrupt:
        pass
    finally:
        pygame.mixer.music.stop()
        pygame.mixer.quit()
```

```python
play_midi("unravel.mid")
```

```python
notes = open_midi("unravel.mid")

print(f"Shape: {notes.shape}")
print(f"Columnas: [tiempo_absoluto, nota, velocidad, canal]")
print(notes[:20])
```
