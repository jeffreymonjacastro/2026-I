# Dataset Structure Diagnostic
Run this BEFORE the main pipeline to understand exactly what files are in each folder.

```python
from pathlib import Path
from collections import defaultdict

DATASET_ROOT = Path('/kaggle/input/datasets/jeffreyamc/oxford-paris-buildings-v2')

for dataset_name in ['oxford', 'paris']:
    ds_root = DATASET_ROOT / dataset_name
    if not ds_root.exists():
        print(f'{dataset_name}: NOT FOUND')
        continue

    print(f'\n{"="*60}')
    print(f'  {dataset_name.upper()}')
    print(f'{"="*60}')

    landmark_dirs = sorted(d for d in ds_root.iterdir() if d.is_dir())

    for lm in landmark_dirs:
        # Count file types
        all_files  = list(lm.iterdir())
        imgs       = [f for f in all_files if f.suffix.lower() in ('.jpg','.jpeg','.png')]
        txt_files  = [f for f in all_files if f.suffix == '.txt']
        query_txts = [f for f in txt_files if '_query' in f.name]
        good_txts  = [f for f in txt_files if '_good'  in f.name]
        ok_txts    = [f for f in txt_files if '_ok'    in f.name]
        junk_txts  = [f for f in txt_files if '_junk'  in f.name]
        other_txts = [f for f in txt_files
                      if not any(t in f.name for t in ('_query','_good','_ok','_junk'))]

        has_gt = len(query_txts) > 0
        tag    = '✅ GT' if has_gt else '⚠️  DISTRACTOR (no query.txt)'

        print(f'\n  [{tag}]  {lm.name}/')
        print(f'    Images       : {len(imgs)}')
        print(f'    *_query.txt  : {len(query_txts)}')
        print(f'    *_good.txt   : {len(good_txts)}')
        print(f'    *_ok.txt     : {len(ok_txts)}')
        print(f'    *_junk.txt   : {len(junk_txts)}')
        if other_txts:
            print(f'    Other .txt   : {[f.name for f in other_txts]}')

        # Show first query file content (so we can see the format)
        if query_txts:
            with open(query_txts[0]) as f:
                print(f'    Sample query : {f.read().strip()}')
```

```python
# Also check if there are files DIRECTLY in oxford/ or paris/ (not in subfolders)
for dataset_name in ['oxford', 'paris']:
    ds_root = DATASET_ROOT / dataset_name
    if not ds_root.exists():
        continue
    direct_files = [f for f in ds_root.iterdir() if f.is_file()]
    print(f'{dataset_name}/ direct files: {[f.name for f in direct_files[:10]]}')
```
