# PyTorch Methods


```python
import torch
import torch.nn as nn
from torch import Tensor
```

```python
# Supongamos que tienes un lote de 32 imágenes, cada una con 3 canales y de 28x28 píxeles.
# Forma del tensor: [batch_size, channels, height, width] -> [32, 3, 28, 28]
x: Tensor = torch.randn(32, 3, 28, 28)

# Creamos la capa Flatten
flatten: nn.Flatten = nn.Flatten()

# Pasamos el tensor por la capa
salida: Tensor = flatten(x)

# La nueva forma será [32, 3 * 28 * 28] = [32, 2352]
print(salida.shape)  # Imprime: torch.Size([32, 2352])
```

```python
def Perceptron():
    return nn.Sequential(
        nn.Flatten(), nn.Linear(32 * 32, 10)  # 2D -> 1D  # classes: 0 - 9
    )
```

```python
# 2. Instanciamos o "creamos" nuestro modelo
modelo = Perceptron()

# 3. Creamos un lote de datos de prueba
# Simularemos 4 imágenes, de 1 canal (escala de grises, como MNIST), y de 32x32 píxeles
# Tamaño: [batch_size=4, channels=1, height=32, width=32]
imagenes_falsas = torch.randn(4, 1, 32, 32)

# 4. Pasamos las imágenes a través del modelo (hacemos una "predicción" o "forward pass")
# Nota: En PyTorch no necesitas llamar a un método ".predict()", solo usas la variable como una función.
salida = modelo(imagenes_falsas)

# 5. Analizamos las formas de los datos
print("Forma de las imágenes de entrada:", imagenes_falsas.shape)
print("Forma de la salida:", salida.shape)
print("\nPuntajes (logits) para la primera imagen:\n", salida[0])
```
