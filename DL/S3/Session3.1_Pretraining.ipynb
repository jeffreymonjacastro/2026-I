# Fine tuning, Transfer learning

No siempre necesitaremos hacer fine-tuning sobre todos los pesos pre-entrenados en Imagenet.

En muchas aplicaciones solo necesitaremos estrenar la fully connected (FC), o un subgrupo de capas del backbone (ResNet, en el ejemplo)

```python
# Esta función fijará los pesos como no entrenables (sin gradientes)
def set_requires_grad(module, flag):
    for p in module.parameters():
        p.requires_grad = flag
```

```python
# Coloca en eval() todas las capas BatchNorm
# Esto evita que se acumulen las estadísticas en las capas que no queremos entrenar
def set_bn_eval(module):
    for m in module.modules():
        if isinstance(m, (nn.BatchNorm1d, nn.BatchNorm2d, nn.BatchNorm3d)):
            m.eval()
```

#### Solo entrenar la FC

```python
model = models.resnet50(weights=models.ResNet50_Weights.IMAGENET1K_V2)
```

```python
num_classes = 20
in_features = model.fc.in_features
model.fc = nn.Linear(in_features, num_classes)
```

```python
# Seleccionamos las capas que queremos congelar
# Para visualizarlas: print(model)
layers = [model.conv1, model.bn1, model.layer1, model.layer2, model.layer3, model.layer4]

for part in layers:
    set_requires_grad(part, False)
    set_bn_eval(part)
```

#### Entrenar layer3, layer4 y la FC

```python
num_classes = 20
in_features = model.fc.in_features
model.fc = nn.Linear(in_features, num_classes)
```

```python
# Seleccionamos las capas que queremos congelar
layers = [model.conv1, model.bn1, model.layer1, model.layer2]
for part in layers:
    set_requires_grad(part, False)
    set_bn_eval(part)
```

```python
# Como buena practica podríamos asegurarnos que las otras capas tengan gradientes
# En general no es necesario
layers = [model.layer3, model.layer4, model.fc]
for part in layers:
    set_requires_grad(part, True)
```

Optimizer solo con parámetros entrenables, asi evitamos que lea parámetros que no va a procesar.

```python
# Solo enviar los parametros que se pueden entrenar
optimizer = torch.optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=5e-4, weight_decay=1e-4)
```

```python
# Una estrategia comun es definir diferentes learning rates para las capas del modelo.
# Recuerda que FC tiene los pesos aleatorios, por lo que la gradiente le debería
# afectar más.
# Por otro lado, los layer3, layer4 ya tienen pesos entrenados: ya pueden generar
# buenas representaciones! así que la gradiente no le debería afectar tanto.
optimizer = torch.optim.AdamW([
    {"params": model.layer3.parameters(), "lr": 3e-4},
    {"params": model.layer4.parameters(), "lr": 3e-4},
    {"params": model.fc.parameters(),     "lr": 1e-3},
], weight_decay=1e-4)
```
